## 1. Applying Custom Functions with `apply()` / `map()`
- `apply()` lets you apply a custom function across rows or columns.
- `map()` is for applying functions element-wise to a Series.
- Both are handy for transformations, calculations, and cleaning data.

```python
import pandas as pd

data = {
    "Name": ["Aarav", "Isha", "Rahul", "Sneha", "Kabir"],
    "Maths": [85, 92, 78, 88, 95],
    "Science": [420, 450, 380, 440, 465]
}

df = pd.DataFrame(data)

print(df)
```

```python
df["Total Marks"] = df[["Maths", "Science"]].apply(
    lambda row: row.sum(),
    axis=1
)

print(df)
```

```python
col_mean = df[["Maths","Science"]].apply(lambda col :col.mean())
col_mean
```

```python
def Grade(s):
  if s>=90:
    return "A"
  elif s>=80:
    return "B"
  elif s>=70:
    return "C"
  else:
    return "D"

df["Maths Grades"] = df["Maths"].apply(Grade)
df
```

```python
df["Name_l"]=df["Name"].map(str.lower)
df
```

```python
df["Bonus Maths"] = df["Maths"].map(lambda v:v+5)
df
```
