## 1. Converting Columns to Datetime (`pd.to_datetime`)
- Use `pd.to_datetime()` to convert strings or numbers into datetime objects.
- Once converted, Pandas recognizes them as dates, enabling powerful operations.

```python
import pandas as pd

sales = pd.DataFrame({
    "Order_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Product": ["Laptop", "Mobile", "Headphones", "Laptop",
                "Mouse", "Mobile", "Keyboard", "Headphones"],
    "Date": ["2026-01-05", "2026-01-12", "2026-01-18", "2026-02-03",
             "2026-02-10", "2026-02-15", "2026-03-02", "2026-03-15"],
    "Quantity": [2, 3, 5, 1, 4, 2, 6, 3],
    "Sales": [110000, 75000, 10000, 55000, 3200, 50000, 9000, 6000],
    "City": ["Mumbai", "Pune", "Delhi", "Mumbai",
             "Nashik", "Pune", "Thane", "Delhi"]
})

print(sales)
```

## 2. Setting DateTime Index & Resampling
- Set a datetime column as the index with `set_index()`.
- Resample time series using `resample('M').mean()` or other rules ('D', 'W', 'Y').
- Useful for aggregating data by time periods.

```python
sales_idx = sales.set_index("Date")
```

```python
sales["Date"] = pd.to_datetime(sales["Date"])

sales_idx = sales.set_index("Date")

sales_idx.resample("ME")["Sales"].sum()
```

```python

# 1️⃣ Create a dataset of website visits (string dates + visit counts)
data = {
    "VisitDate": [
        "2026-01-05", "2026-01-12", "2026-01-20",
        "2026-02-03", "2026-02-10", "2026-02-18",
        "2026-03-02", "2026-03-09", "2026-03-20",
        "2026-04-05", "2026-04-15", "2026-04-25"
    ],
    "Visits": [120, 150, 180, 200, 220, 250, 300, 280, 350, 400, 420, 450]
}

df = pd.DataFrame(data)


# 3️⃣ Set VisitDate as the index
# Write your code here

# 4️⃣ Resample to get total visitors per month
# Write your code here

# 5️⃣ Resample to get average visitors per week
# Write your code here
```

```python
# 2️⃣ Convert VisitDate to datetime
df["VisitDate"] = pd.to_datetime(df["VisitDate"])
```

```python
# 3️⃣ Set VisitDate as the index
df = df.set_index("VisitDate")
```

```python
# 4️⃣ Resample to get total visitors per month
monthly_visits = df.resample("ME")["Visits"].sum()

print("Monthly Total Visits:")
print(monthly_visits)
```

```python
# 5️⃣ Resample to get average visitors per week
weekly_visits = df.resample("W")["Visits"].mean()

print("\nWeekly Average Visits:")
print(weekly_visits)
```
