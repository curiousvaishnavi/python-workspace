## 1. Using `concat()` for Stacking Data
- `pd.concat()` combines DataFrames vertically (rows) or horizontally (columns).
- Use `axis=0` for vertical stacking (default), `axis=1` for horizontal.
- `ignore_index=True` can reset the index after concatenation.

```python

# Dataset 1: Product details
import pandas as pd

data1 = {
    "Product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Price": [55000, 25000, 2000, 1500, 800],
    "Stock": [10, 25, 50, 30, 45]
}

df1 = pd.DataFrame(data1)

print(df1)

# Dataset 2: More products (same columns as df1 → good for vertical stacking)
data2 = {
    "Product": ["Monitor", "Printer", "Tablet", "Webcam", "Speaker"],
    "Price": [12000, 8000, 18000, 2500, 3000],
    "Stock": [15, 20, 12, 35, 40]
}

df2 = pd.DataFrame(data2)

print(df2)

# Dataset 3: Additional info (different columns → good for horizontal stacking)
data3 = {
    "Rating": [4.5, 4.3, 4.2, 4.4, 4.1],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Accessories"]
}

df3 = pd.DataFrame(data3)

print(df3)
```

```python
pd.concat([df1,df2],ignore_index=True)
```

```python
pd.concat([df1,df2],axis=1)
```

```python
pd.concat([df1,df3],axis=1)
```

## 2. Using `merge()` for SQL-Style Joins
- `merge()` is similar to SQL joins.
- Specify `on` or `left_on`/`right_on` to choose key columns.
- Join types: `inner` (default), `left`, `right`, `outer`.

```python
import pandas as pd

# Product data
products = {
    "Product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Price": [55000, 25000, 2000, 1500, 800],
    "Stock": [10, 25, 50, 30, 45]
}

# Dictionary → DataFrame
products = pd.DataFrame(products)

# Sales DataFrame
sales_data = {
    "Order_ID": [101, 102, 103, 104, 105],
    "Product": ["Laptop", "Mobile", "Laptop", "Headphones", "Mouse"],
    "Sales": [55000, 25000, 60000, 4000, 1200],
    "Quantity": [1, 2, 1, 2, 2]
}

df = pd.DataFrame(sales_data)
```

```python
pd.merge(products,df)
```

```python
pd.merge(products, df, on="Product", how="left")
```

```python
pd.merge(products, df, on="Product", how="right")
```

```python
pd.merge(products, df, on="Product", how="outer")
```

```python
pd.merge(products, df, on="Product", how="inner")
```

## 3. Using `join()` with Index Alignment
- `join()` joins DataFrames on their index or on a key column.
- By default, joins on index; can specify `on='col'`.
- Useful for adding columns to an existing DataFrame.

## 4. Practical Examples
- Combine sales and revenue data to see performance.
- Merge students with marks to link records.
- Practice with realistic datasets to understand alignment and missing data handling.

```python
import pandas as pd

sales = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Headphones", "Mouse"],
    "Units_Sold": [10, 20, 30, 40]
})

revenue = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Headphones", "Mouse"],
    "Revenue": [550000, 500000, 120000, 32000]
})

performance = pd.merge(sales, revenue, on="Product",how = "outer")

print(performance)
```

## Mini Practice
1. Create three DataFrames: Products, Sales, and Revenue.
2. Concatenate them vertically and horizontally.
3. Merge Sales and Revenue using different join types.
4. Join Products with Sales based on index or a key.
5. Analyze missing values after joins.

```python

# 1. Create three DataFrames: Products, Sales, and Revenue
import pandas as pd

products = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Price": [55000, 25000, 2000, 1500, 800],
    "Stock": [10, 25, 50, 30, 45]
})

print(products)

sales = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Units_Sold": [8, 20, 35, 15, 30],
    "City": ["Mumbai", "Pune", "Delhi", "Mumbai", "Nashik"]
})

print(sales)

revenue = pd.DataFrame({
    "Product": ["Laptop", "Mobile", "Headphones", "Keyboard", "Mouse"],
    "Revenue": [440000, 500000, 70000, 22500, 24000],
    "Profit": [80000, 100000, 15000, 5000, 6000]
})

print(revenue)

```

```python
# 2. Concatenate them vertically and horizontally


pd.concat([products,sales,revenue],axis = 0)
```

```python
pd.concat([products,sales,revenue],axis = 1)
```

```python
# 3. Merge Sales and Revenue using different join types (inner, left, right, outer)

pd.merge(sales,revenue,on ="Product",how = "left" )
```

```python
# 3. Merge Sales and Revenue using different join types (inner, left, right, outer)

pd.merge(sales,revenue,on ="Product",how = "right" )
```

```python
# 4. Join Products with Sales based on index or a key


result = products.join(
    sales.set_index("Product"),
    on="Product"
)

print(result)
```

```python
# 5. Analyze missing values after joins
# Use isna(), info(), or describe() to explore gaps


result.describe()
```
